# Explore a FortiGate from a notebook

Run the cells one by one (Shift+Enter). Every call is carried out by ZapTerm on a
real terminal tab — watch it there — and the output comes back here, so you can
try parsing it straight away.

**Before you start**

1. In ZapTerm: **Tools → Allow notebook connection…**
2. Open the device you want to look at in a ZapTerm tab (or use `zapterm.open()` below).

🔒 The notebook connection starts **read-only**: only `get` / `show` commands
reach the device, anything else is listed in ZapTerm as "would run". The cells
below only use `get` commands.

```python
import zapterm

zapterm.connect()   # finds the running ZapTerm on its own
zapterm.tabs()      # the tabs open in ZapTerm right now
```

## Pick the device

Use the name shown on the ZapTerm tab. To open a saved session instead, use
`zapterm.open("saved session name")`.

```python
DEVICE = "FGT-60F-01"   # ← the name of an open ZapTerm tab

fw = zapterm.tab(DEVICE)    # or: fw = zapterm.open(DEVICE)
fw
```

```python
status = fw.run("get system status")
print(status)
```

```python
summary = fw.run("get vpn ipsec tunnel summary")
print(summary)
```

## Work on the output

`summary` is a plain string: try your parsing here until it is right, then copy
it into a script (see `sase_reset.py`).

```python
import re

tunnels = {
    name: (int(total), int(up))
    for name, total, up in re.findall(
        r"'([^']+)'.*?selectors\(total,up\):\s*(\d+)/(\d+)", summary
    )
}
down = [name for name, (total, up) in tunnels.items() if total == 0 or up < total]
print("tunnels:", tunnels)
print("down:   ", down or "none")
```
